**Listing directories inside dataset**

```python
dataset_path = "/content/drive/MyDrive/Deep-Learning-Image-Classification/Dataset"

print(os.listdir(dataset_path))

train_path = dataset_path + "/train"
test_path = dataset_path + "/test"
val_path = dataset_path + "/val"


print(os.listdir(train_path))
```

**Counting no of images belonging to each class**

```python
import os

for class_name in os.listdir(train_path):
    class_path = os.path.join(train_path, class_name)

    if os.path.isdir(class_path):
        count = len(os.listdir(class_path))
        print(class_name, count)

print("\n")

for class_name in os.listdir(val_path):
    class_path = os.path.join(val_path, class_name)

    if os.path.isdir(class_path):
        count = len(os.listdir(class_path))
        print(class_name, count)

print("\n")
for class_name in os.listdir(test_path):
    class_path = os.path.join(test_path, class_name)

    if os.path.isdir(class_path):
        count = len(os.listdir(class_path))
        print(class_name, count)
```

**Visualizing a sample image and size**

```python
import matplotlib.pyplot as plt
from PIL import Image

sample_image = os.path.join(
train_path,
"mountain",
os.listdir(os.path.join(train_path, "mountain"))[150]
)
img = Image.open(sample_image)
plt.imshow(img)
plt.axis('off')
plt.show()
print(img.size)
```

**Counting the total of images**

```python
total_images = 0

for class_name in os.listdir(train_path):

    folder = os.path.join(train_path, class_name)

    if os.path.isdir(folder):

        count = len(os.listdir(folder))

        total_images += count

print("Total Training Images:", total_images)
total_images = 0
print("\n")
for class_name in os.listdir(val_path):

    folder = os.path.join(val_path, class_name)

    if os.path.isdir(folder):

        count = len(os.listdir(folder))

        total_images += count
print("Total validating Images:", total_images)
total_images = 0

print("\n")
for class_name in os.listdir(test_path):

    folder = os.path.join(test_path, class_name)

    if os.path.isdir(folder):

        count = len(os.listdir(folder))

        total_images += count
print("Total Test Images:", total_images)
```
